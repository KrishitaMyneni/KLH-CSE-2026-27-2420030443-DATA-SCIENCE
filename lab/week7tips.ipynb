```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
df = sns.load_dataset('tips')

plt.figure(figsize=(6,4))
sns.histplot(df['total_bill'], bins=20, kde=True)
plt.title("Histogram of Total Bill")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(x=df['total_bill'])
plt.title("Boxplot of Total Bill")
plt.show()
```

```python
day_counts = df['day'].value_counts()
plt.figure(figsize=(6,6))
plt.pie(day_counts, labels=day_counts.index, autopct='%1.1f%%')
plt.title("Day Distribution (pie chart)")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.scatterplot(x=df['total_bill'], y=df['tip'], hue=df['sex'])
plt.title("Scatter Plot: Total Bill vs Tip")
plt.show()

plt.figure(figsize=(6,4))
plt.plot(df['total_bill'])
plt.title("Line chart of Total Bill")
plt.xlabel("Sample index")
plt.ylabel("Total Bill")
plt.show()

plt.figure(figsize=(6,4))
sns.barplot(x=df['day'], y=df['total_bill'])
plt.title("Bar Chart: Mean Total Bill by Day")
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(df[['total_bill', 'tip', 'size']].corr(), annot=True, cmap="coolwarm")
plt.title("Heatmap of Feature correlations")
plt.show()

plt.figure(figsize=(6,4))
plt.scatter(df['total_bill'], df['tip'], df['size']*20, alpha=0.5)
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.show()

sns.pairplot(df[['total_bill', 'tip', 'size']])
plt.suptitle("Pair Plot of Tips Features", y=1.02)
plt.show()
```

```python
# violin plot
plt.figure(figsize=(6,4))
sns.violinplot(x=df['day'], y=df['tip'])
plt.title("Violin Plot: Tip by Day")
plt.show()
```
