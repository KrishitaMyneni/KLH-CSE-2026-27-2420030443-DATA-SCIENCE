```python
#even_odd
def even_odd(num):
    if num % 2 == 0:
        return "Even"
    else:
        return "Odd"
num=int(input("Enter a number: "))
result=even_odd(num)
print("The number", num, "is", result)
```

```python
#square
def find_square(numb):
    return numb * numb
numb=int(input())
square=find_square(numb)
print("the square of", numb, "is", square)


```

```python
#while loop
i=1
while i<6:
    print(i)
    i+=1
#while loop with break statement
i=1
while i<6:
    print(i)
    if i==3:
        break
    i+=1
#while loop with continue statement
i=0
while i<6:
    i+=1
    if i==3:
        continue
    print(i)
```
